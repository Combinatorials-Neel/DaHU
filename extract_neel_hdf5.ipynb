 ### EXPORT SPECTRA INTO XY FILES
 ### ALL IMAGES WILL BE EXTRACTED INTO PNG FILES

```python
# XRD
with h5py.File(hdf5_path, "r") as hdf5_file:
    hdf5_path = Path(hdf5_path)
    export_path = make_exports_folder(hdf5_path)
    print(export_path)
    for name, group in hdf5_file.items():
        try:
            if group.attrs["HT_type"] == "xrd":
                xrd_export_spectra_to_files(group, export_path, save_image=False)
            if group.attrs["HT_type"] == "moke":
                hdf5_export_moke_loops(group, export_path)
            if group.attrs["HT_type"] == "sem":
                hdf5_export_sem_images(group, export_path, format="png")
        except KeyError:
            continue


    hdf5_file.flush()
```
