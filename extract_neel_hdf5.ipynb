```python
# XRD
with h5py.File(hdf5_path, "r") as hdf5_file:
    hdf5_path = Path(hdf5_path)
    export_path = make_exports_folder(hdf5_path)
    for name, group in hdf5_file.items():
        try:
            if group.attrs["HT_type"] == "xrd":
                xrd_export_spectra_to_files(group, export_path, save_image=False)
        except KeyError:
            continue


    hdf5_file.flush()
```
